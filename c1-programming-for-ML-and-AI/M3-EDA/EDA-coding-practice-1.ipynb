Instructions

You are working as a typist in the HR department of a company. Your job was to record the email addresses of candidates during onboarding. Unfortunately, the '@' and the '.' keys on your keyboard were malfunctioning, and as a result, some email addresses were entered without the '@' or the '.' symbols. Now, your manager has asked you to identify all such email entries that are potentially invalid due to the missing '@' and '.' symbols. These candidates will be contacted again to re-submit their correct email addresses.



For the purpose of this exercise, assume that a valid email ID is one which adheres to the following rules

It is a non-empty string
It must contain both an '@' symbol and a '.' symbol
The '.' symbol must come after the '@' symbol with at least one character in between them


Write a Python function identify_invalid_email(record) that scans the provided data entry (a dictionary with the candidate name and their email ID) and replaces their email ID with a string 'invalid' if it is invalid. If not, the email ID is unmodified. Return the dictionary after these tasks.



Input Format

The input is a dictionary that carries the candidate's name and email ID in the following format:
The keys are 'Candidate Name' and 'Email ID'
The values are their name (str) and their email ID (str) respectively


Output Format

The output is a dictionary that carries the candidate's name and updated email ID in the following format:
The keys are 'Candidate Name' and 'Email ID'
The values are their name (str) and their updated email ID (str) respectively
The updated email ID would be the same as the input email ID if it is valid, but would be 'invalid' if not


Constraints

The input dictionary is not empty and contains the keys 'Candidate Name' and 'Email ID'
The candidate name is a non-empty string
There are no whitespaces in the email ID
The email ID contains at most one '@' symbol and at most one '.' symbol
The email ID does not start or end with an '@' symbol or a '.' symbol


Example Cases



Testcase 1



Input

{'Candidate Name': 'Nikhil', 'Email ID': 'nikhil98singhdomain.com'}



Expected Output

{'Candidate Name': 'Nikhil', 'Email ID': 'invalid'}



Testcase 2



Input

{'Candidate Name': 'Dravid', 'Email ID': 'dravidlee@domaincom'}



Expected Output

{'Candidate Name': 'Dravid', 'Email ID': 'invalid'}

```python
# Libraries (do not edit)
from ast import literal_eval
import pandas as pd

# Function to identify invalid email ID
def identify_invalid_email(record):
    # Code here
    email = record['Email ID']
    
    # Check if email is non-empty and contains both '@' and '.'
    if email and '@' in email and '.' in email:
        # Check if '.' comes after '@' with at least one character in between
        at_index = email.index('@')
        dot_index = email.index('.')
        if dot_index > at_index + 1:
            # Email is valid, keep as is
            return record
    
    # Email is invalid
    record['Email ID'] = 'invalid'
    return record

# Input and output processing  (do not edit)
print(identify_invalid_email(literal_eval(input())))
```
