```python
import os
import pandas as pd

# Change working directory to the specified path
os.chdir('/content/drive/MyDrive/epgp-aai-aai-c1-ma-eda')
file_list = os.listdir()

df = None

# Iterate through each file in the directory
for file_name in file_list:
    try:
        file_path = os.path.join(os.getcwd(), file_name)
        # Read CSV file into DataFrame
        monthly_df = pd.read_csv(file_path)
        
        # Convert 'timestamp' column to datetime format
        monthly_df['timestamp'] = pd.to_datetime(monthly_df['timestamp'])
        # Extract date and hour from timestamp
        monthly_df['date'] = monthly_df['timestamp'].dt.date
        monthly_df['hour'] = monthly_df['timestamp'].dt.hour

        # Sample 5% of data for each date-hour group
        sampled_data = monthly_df.groupby(['date', 'hour']).apply(lambda x: x.sample(frac=0.05)).reset_index(drop=True)
        
        # Concatenate sampled data from each file
        if df is None:
            df = sampled_data
        else:
            df = pd.concat([df, sampled_data], ignore_index=True)
    except Exception as e:
        # Print error message if file reading fails
        print(f"Error reading file {file_name}: {e}")

# Display the final concatenated DataFrame
display(df)
```
